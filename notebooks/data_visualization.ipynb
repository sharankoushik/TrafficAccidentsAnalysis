# Traffic Accident Data Visualization — Issue #5

This notebook creates the required visualizations for Issue #5 directly from the raw traffic accident dataset. It produces yearly/monthly trends, severity, location, weather, road condition, vehicle type and a weather-vs-road heatmap.

**Geographic map:** the supplied dataset does not contain latitude/longitude columns, so a hotspot map is documented as not applicable rather than using invented coordinates.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DATA_PATH = ROOT / 'data' / 'traffic_accidents_raw.csv'
OUT = ROOT / 'results' / 'figures'
OUT.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(DATA_PATH)
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df.head()
```

```python
df.shape, df.columns.tolist()
```

## 1. Accidents by Year

```python
year_counts = df['Year'].value_counts().sort_index()
fig, ax = plt.subplots(figsize=(10,6))
ax.plot(year_counts.index, year_counts.values, marker='o')
ax.set_title('Traffic Accidents by Year')
ax.set_xlabel('Year')
ax.set_ylabel('Number of Accidents')
ax.grid(True, alpha=0.25)
fig.tight_layout()
fig.savefig(OUT / 'accidents_by_year.png', dpi=200, bbox_inches='tight')
plt.show()
```

## 2. Accidents by Month

```python
month_counts = df['Month'].value_counts().sort_index()
fig, ax = plt.subplots(figsize=(10,6))
ax.plot(month_counts.index, month_counts.values, marker='o')
ax.set_title('Traffic Accidents by Month')
ax.set_xlabel('Month')
ax.set_ylabel('Number of Accidents')
ax.set_xticks(range(1,13))
fig.tight_layout()
fig.savefig(OUT / 'accidents_by_month.png', dpi=200, bbox_inches='tight')
plt.show()
```

## 3. Accidents by Severity

```python
severity = df['Accident_Severity'].value_counts()
fig, ax = plt.subplots(figsize=(9,6))
severity.plot(kind='bar', ax=ax)
ax.set_title('Traffic Accidents by Severity')
ax.set_xlabel('Accident Severity')
ax.set_ylabel('Number of Accidents')
fig.tight_layout()
fig.savefig(OUT / 'accidents_by_severity.png', dpi=200, bbox_inches='tight')
plt.show()
```

## 4. Accidents by Location

```python
location = df['City'].value_counts().head(20)
fig, ax = plt.subplots(figsize=(11,7))
location.sort_values().plot(kind='barh', ax=ax)
ax.set_title('Top 20 Cities by Number of Accidents')
ax.set_xlabel('Number of Accidents')
ax.set_ylabel('City')
fig.tight_layout()
fig.savefig(OUT / 'accidents_by_location.png', dpi=200, bbox_inches='tight')
plt.show()
```

## 5. Accidents by Weather Condition

```python
weather = df['Weather_Condition'].value_counts()
fig, ax = plt.subplots(figsize=(9,6))
weather.plot(kind='bar', ax=ax)
ax.set_title('Traffic Accidents by Weather Condition')
ax.set_xlabel('Weather Condition')
ax.set_ylabel('Number of Accidents')
fig.tight_layout()
fig.savefig(OUT / 'accidents_by_weather.png', dpi=200, bbox_inches='tight')
plt.show()
```

## 6. Accidents by Road Condition

```python
road = df['Road_Condition'].value_counts()
fig, ax = plt.subplots(figsize=(9,6))
road.plot(kind='bar', ax=ax)
ax.set_title('Traffic Accidents by Road Condition')
ax.set_xlabel('Road Condition')
ax.set_ylabel('Number of Accidents')
fig.tight_layout()
fig.savefig(OUT / 'accidents_by_road_condition.png', dpi=200, bbox_inches='tight')
plt.show()
```

## 7. Accidents by Vehicle Type

```python
vehicle = df['Vehicle_Type'].value_counts()
fig, ax = plt.subplots(figsize=(9,6))
vehicle.plot(kind='bar', ax=ax)
ax.set_title('Traffic Accidents by Vehicle Type')
ax.set_xlabel('Vehicle Type')
ax.set_ylabel('Number of Accidents')
fig.tight_layout()
fig.savefig(OUT / 'accidents_by_vehicle_type.png', dpi=200, bbox_inches='tight')
plt.show()
```

## 8. Accident Heatmap

```python
heatmap = pd.crosstab(df['Weather_Condition'], df['Road_Condition'])
fig, ax = plt.subplots(figsize=(10,6))
im = ax.imshow(heatmap.values, aspect='auto')
ax.set_title('Accident Count: Weather Condition vs Road Condition')
ax.set_xlabel('Road Condition')
ax.set_ylabel('Weather Condition')
ax.set_xticks(range(len(heatmap.columns)))
ax.set_xticklabels(heatmap.columns, rotation=30, ha='right')
ax.set_yticks(range(len(heatmap.index)))
ax.set_yticklabels(heatmap.index)
for i in range(heatmap.shape[0]):
    for j in range(heatmap.shape[1]):
        ax.text(j, i, heatmap.iloc[i,j], ha='center', va='center')
fig.colorbar(im, ax=ax, label='Number of Accidents')
fig.tight_layout()
fig.savefig(OUT / 'accident_heatmap.png', dpi=200, bbox_inches='tight')
plt.show()
```
